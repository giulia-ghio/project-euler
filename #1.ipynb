If we list all the natural numbers below 10 that are multiples of 3 or 5, we get 3, 5, 6 and 9 The sum of these multiples is 23.
Find the sum of all the multiples of 3 or 5 below 1000.

```python
#Simply list all multiples
multiples_of_3 = {3*i for i in range(1,1000) if 3*i < 1000 }
multiples_of_5 = {5*i for i in range(1,1000) if 5*i < 1000 }

#Union of both sets to get all of the multiples. The reason I chose sets is to remove duplicates, as multiples of 15 will be in both sets.
#This could've been done very easily with a list, but I wanted to show the use of sets.
all_multiples = multiples_of_3.union(multiples_of_5)

#Simply sum all of the multiples
sum(all_multiples)

print("The sum of all multiples of 3 and 5 below 1000 is:", sum(all_multiples))
```
