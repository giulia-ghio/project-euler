```python
#I will try to solve the problem by an alternative method, using the closed form of the Fibonacci sequence, known as Binet's formula.
#This is mostly for fun, but it also greatly reduces the time complexity of the solution, since we can calculate the Fibonacci numbers in O(1) time.

phi = 1/2 * (1+5**0.5)
psi = 1/2 * (1-5**0.5)
F = lambda n: 1 / 5 ** 0.5 * (phi**n - psi**n)

#Applying some basic algebra, it can be shown that, if F(n) is even, then n is a multiple of 3.
#Moreso, it is possible to find the largest index n such that F(n) is smaller than a given number x, by using the following formula:
#n = floor(log_phi(5 ** 0.5 * (x + 1/2)))

import numpy as np

log_phi = lambda x: np.log(x) / np.log(phi)
n_max = lambda x: np.floor(log_phi(5 ** 0.5 * (x + 1/2)))
```

```python
#Finding the largest index n such that F(n) is smaller than 4,000,000
N = n_max(4000000)
print(N)

#Let's just test applying that to see if the limit checks out.
print(F(N-1),F(N),F(N+1))
#This is correct!

#Now, for the sum of all even Fibonacci numbers:
S = sum(F(n) for n in range(3, int(N+3), 3))

print(int(S))
#The rounding errors are so small that we can just round the result to the nearest integer and it will be correct.
#Rigorously, I'd need to use some other algorithm to estimate the rounding errors, but I don't think it's necessary in this case.
```
